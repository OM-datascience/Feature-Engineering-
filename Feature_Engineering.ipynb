**Q1. What is a parameter?**

Ans1. In machine learning, a parameter is basically a value that the model learns on its own from the training data, rather than something you manually set beforehand, like the weights and bias in a linear regression model. These are the actual numbers that get adjusted during training so the model can make better predictions, and once training is done, they represent what the model has actually "learned" from the data.

**Q2. What is correlation? What does negative correlation mean?**

Ans2. Correlation is a statistical measure that tells you how strongly two variables are related to each other, and it's usually expressed as a number between -1 and 1. A negative correlation means that as one variable increases, the other tends to decrease, so they move in opposite directions, like how the number of hours spent on ice or snow exposure might go down as outdoor temperature goes up. The closer that negative value gets to -1, the stronger and more consistent this inverse relationship is.

**Q3. Define Machine Learning. What are the main components in Machine Learning?**

Ans3. Machine Learning is basically a branch of AI where a system learns patterns from data and uses that to make predictions or decisions, without being explicitly programmed with hardcoded rules for every scenario. The main components involved are the data itself, which the model learns from, the model or algorithm that captures patterns in that data, the features, which are the input variables used for learning, the labels or target values in supervised learning, and the loss function, which measures how far off the model's predictions are so it knows how to improve.

**Q4. How does loss value help in determining whether the model is good or not?**

Ans4. The loss value is a number that tells you how far off the model's predictions are from the actual correct values, so a lower loss generally means the model is doing a better job. During training, the model keeps adjusting its parameters specifically to reduce this loss, and by tracking how it changes over time, you can tell whether the model is actually learning or just stuck, or even overfitting if the training loss keeps dropping but performance on new data doesn't improve.

**Q5. What are continuous and categorical variables?**

Ans5. Continuous variables are numeric values that can take any value within a range, and they can be measured to any level of precision, things like height, weight, or temperature. Categorical variables, on the other hand, represent distinct groups or categories rather than numbers with real numeric meaning, like gender, city names, or product types, and they can be further split into nominal, where there's no order like colors, and ordinal, where there is a natural order like small, medium, large.

**Q6. How do we handle categorical variables in Machine Learning? What are the common techniques?**

Ans6. Since most machine learning algorithms only work with numbers, categorical variables need to be converted into a numeric format before they can be used. The most common techniques are label encoding, where each category just gets assigned a number, one hot encoding, where each category becomes its own binary column of 0s and 1s, and ordinal encoding, which is similar to label encoding but specifically respects a natural order among the categories, like rating levels of low, medium, and high.

**Q7. What do you mean by training and testing a dataset?**

Ans7. Training a dataset means using a portion of your data to actually teach the model, letting it learn the underlying patterns and relationships between the features and the target. Testing means holding back a separate portion of data that the model has never seen during training, and using that afterward to check how well the model actually performs on new, unseen data, since that's the real measure of whether it generalizes well or just memorized the training data.

**Q8. What is sklearn.preprocessing?**

Ans8. sklearn.preprocessing is a module in the Scikit-learn library that provides tools for preparing and transforming raw data before feeding it into a machine learning model. It includes things like scalers for standardizing or normalizing numeric features, and encoders for converting categorical variables into numeric form, all steps that usually need to happen before the actual model training begins.

**Q9. What is a Test set?**

Ans9. A test set is the portion of your dataset that's kept completely separate from training and only used at the very end, to evaluate how well the trained model performs on data it has never seen before. It acts as a fair, unbiased check on the model's real world performance, which is important because checking performance only on training data can give an overly optimistic and misleading picture.

**Q10. How do we split data for model fitting (training and testing) in Python? How do you approach a Machine Learning problem?**

Ans10. In Python, splitting data for training and testing is usually done using the `train_test_split()` function from `sklearn.model_selection`, where you pass in your features and target, along with a test size like 0.2 for an 80-20 split. As for approaching a machine learning problem in general, it usually starts with understanding the problem and the data through EDA, then cleaning and preprocessing the data, splitting it into training and test sets, choosing and training an appropriate model, evaluating its performance on the test set, and finally tuning it further if the results aren't good enough.

**Q11. Why do we have to perform EDA before fitting a model to the data?**

Ans11. EDA, or exploratory data analysis, helps you actually understand what you're working with before jumping straight into modeling, things like spotting missing values, outliers, wrong data types, or strong relationships between variables. Skipping this step and going straight to model fitting often leads to problems down the line, like a model that performs badly because of messy data or unnoticed issues that could've been caught early with some basic visualization and summary statistics.

**Q12. What is correlation?**

Ans12. Correlation measures how strongly two variables move together, expressed as a value between -1 and 1, where values close to 1 mean a strong positive relationship, values close to -1 mean a strong negative relationship, and values near 0 mean there's little to no linear relationship between them at all.

**Q13. What does negative correlation mean?**

Ans13. A negative correlation means the two variables move in opposite directions, so as one goes up, the other tends to go down, like how the amount of time spent studying might be negatively correlated with the number of mistakes made on an exam.

**Q14. How can you find correlation between variables in Python?**

Ans14. In Python, if you have your data in a Pandas DataFrame, you can just call `df.corr()` and it'll return a correlation matrix showing the pairwise correlation between all numeric columns. You can also visualize this more clearly by passing that same matrix into a Seaborn heatmap, which makes it a lot easier to spot strong relationships at a glance.

**Q15. What is causation? Explain difference between correlation and causation with an example.**

Ans15. Causation means one variable directly causes a change in another, there's an actual cause and effect relationship, unlike correlation, which just tells you two things happen to move together without necessarily one causing the other. A classic example is how ice cream sales and drowning incidents are correlated because both increase during summer, but eating ice cream obviously doesn't cause drowning, the real underlying cause is just the hot weather affecting both independently.

**Q16. What is an Optimizer? What are different types of optimizers? Explain each with an example.**

Ans16. An optimizer is the algorithm responsible for adjusting a model's parameters during training in order to minimize the loss function, basically it's what actually drives the learning process. Gradient Descent is the most basic type, where parameters are updated step by step in the direction that reduces loss the most, Stochastic Gradient Descent updates parameters using just one training example at a time instead of the whole dataset, making it faster but noisier, and Adam is a more advanced optimizer that combines ideas from both momentum and adaptive learning rates, which is why it's commonly used as a default choice in deep learning today.

**Q17. What is sklearn.linear_model?**

Ans17. sklearn.linear_model is a module within Scikit-learn that contains implementations of various linear models, like LinearRegression for regression tasks and LogisticRegression for classification tasks. It gives you ready to use classes where you can just fit your data and get predictions, without needing to code the underlying math yourself.

**Q18. What does model.fit() do? What arguments must be given?**

Ans18. model.fit() is the method that actually trains the model, it takes in your training data and adjusts the model's internal parameters so it learns the relationship between the inputs and outputs. In supervised learning, you typically need to pass in two arguments, the feature matrix X, containing your input variables, and the target array y, containing the actual values or labels you want the model to learn to predict.

**Q19. What does model.predict() do? What arguments must be given?**

Ans19. model.predict() is used after the model has already been trained, and it generates predictions for new input data based on what the model has learned. It only needs one argument, the feature matrix X for the data you want predictions on, and it returns the predicted output values for each row in that input.

**Q20. What are continuous and categorical variables?**

Ans20. Continuous variables are numeric and can take any value within a range, like age or income, while categorical variables represent distinct categories or labels rather than numbers with real magnitude, like marital status or product type.

**Q21. What is feature scaling? How does it help in Machine Learning?**

Ans21. Feature scaling is the process of adjusting numeric features so they're all on a similar scale, since features with very different ranges, like income in thousands versus age in tens, can cause certain algorithms to give more importance to the larger valued feature simply because of its scale rather than its actual importance. It helps a lot with distance based algorithms like KNN or SVM, and also speeds up convergence for algorithms that rely on gradient descent, since the optimizer doesn't have to deal with wildly different scales across features.

**Q22. How do we perform scaling in Python?**

Ans22. Scaling in Python is usually done using classes from `sklearn.preprocessing`, like `StandardScaler`, which centers the data around a mean of 0 with a standard deviation of 1, or `MinMaxScaler`, which squishes all values into a fixed range like 0 to 1. You typically fit the scaler on your training data first, and then use that same fitted scaler to transform both the training and test sets consistently.

**Q23. What is sklearn.preprocessing?**

Ans23. It's the Scikit-learn module dedicated to preparing raw data for machine learning, offering tools for scaling numeric features, encoding categorical variables, and handling other transformations that typically need to happen before a model can actually be trained on the data.

**Q24. How do we split data for model fitting (training and testing) in Python?**

Ans24. You use the `train_test_split()` function from `sklearn.model_selection`, passing in your features and target variable along with a chosen test size, like 20% of the data reserved for testing, and it randomly splits everything into separate training and testing sets ready to be used for fitting and evaluating a model.

**Q25. Explain data encoding?**

Ans25. Data encoding is the process of converting categorical, non-numeric data into a numeric format that a machine learning model can actually understand and work with. Some common methods include label encoding, which assigns a unique number to each category, and one hot encoding, which creates separate binary columns for each category, both of which serve the same overall purpose of making text based or categorical data usable in mathematical models.
